```python
# Student Performance Data Analysis System

### Data Analysis Essentials Cornerstone Project
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student_ID": range(1, 21),

    "Name": [
        "Arjun", "Rahul", "Priya", "Sneha", "Kiran",
        "Anjali", "Vijay", "Ravi", "Meena", "Suresh",
        "Divya", "Rohan", "Pooja", "Nikhil", "Swathi",
        "Varun", "Keerthi", "Manoj", "Lakshmi", "Aditya"
    ],

    "Gender": [
        "Male", "Male", "Female", "Female", "Male",
        "Female", "Male", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],

    "Age": [
        18, 19, 18, 19, 18,
        18, 19, 18, 19, 18,
        18, 19, 18, 18, 19,
        18, 19, 18, 19, 18
    ],

    "Maths": [
        85, 72, 91, 65, 78,
        88, 56, 74, 95, 68,
        82, 90, 61, 76, 87,
        69, 93, 58, 80, 84
    ],

    "Physics": [
        78, 68, 89, 62, 75,
        84, 59, 70, 92, 64,
        79, 86, 60, 72, 83,
        65, 90, 55, 77, 81
    ],

    "Chemistry": [
        82, 75, 94, 68, 80,
        86, 54, 71, 96, 70,
        85, 88, 63, 78, 89,
        67, 91, 52, 79, 83
    ],

    "English": [
        88, 80, 92, 72, 85,
        90, 62, 78, 94, 75,
        86, 91, 70, 82, 88,
        73, 95, 65, 84, 87
    ],

    "Computer_Science": [
        92, 78, 96, 70, 88,
        94, 60, 82, 98, 73,
        90, 95, 68, 85, 93,
        76, 97, 63, 89, 91
    ],

    "Attendance": [
        92, 85, 96, 78, 88,
        94, 72, 83, 98, 80,
        91, 95, 76, 87, 93,
        79, 97, 70, 89, 92
    ]
}

df = pd.DataFrame(data)

df.head()
```

```python
print("Number of Students:", len(df))
print("Number of Columns:", len(df.columns))

df.info()
```

```python
df.describe()
```

```python
subjects = [
    "Maths",
    "Physics",
    "Chemistry",
    "English",
    "Computer_Science"
]

df["Average"] = df[subjects].mean(axis=1)

df[["Name", "Average"]].head()
```

```python
def calculate_grade(average):

    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


df["Grade"] = df["Average"].apply(calculate_grade)

df[["Name", "Average", "Grade"]]
```

```python
sorted_students = df.sort_values(
    by="Average",
    ascending=False
)

sorted_students[
    ["Name", "Average", "Grade"]
].head(10)
```

```python
topper = df.loc[df["Average"].idxmax()]

print("Topper Name:", topper["Name"])
print("Topper Average:", round(topper["Average"], 2))
print("Topper Grade:", topper["Grade"])
```

```python
low_performers = df[df["Average"] < 60]

low_performers[
    ["Name", "Average", "Grade"]
]
```

```python
low_attendance = df[df["Attendance"] < 75]

low_attendance[
    ["Name", "Attendance"]
]
```

```python
subject_averages = df[subjects].mean()
print(subject_averages)

best_subject = subject_averages.idxmax()
print("Best Performing Subject:", best_subject)

weakest_subject = subject_averages.idxmin()
print("Weakest Performing Subject:", weakest_subject)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    subject_averages.index,
    subject_averages.values
)

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df["Name"],
    df["Average"]
)

plt.title("Student Average Performance")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
grade_distribution = df["Grade"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    grade_distribution.values,
    labels=grade_distribution.index,
    autopct="%1.1f%%"
)

plt.title("Grade Distribution")

plt.show()
```

```python
total_students = len(df)
overall_average = df["Average"].mean()
topper_name = topper["Name"]
topper_average = topper["Average"]
at_risk_students = len(
    df[(df["Average"] < 60) | (df["Attendance"] < 75)]
)

report = f"""
STUDENT PERFORMANCE DATA ANALYSIS REPORT
=========================================

Total Students: {total_students}

Overall Average: {overall_average:.2f}

Topper Name: {topper_name}

Topper Average: {topper_average:.2f}

Best Performing Subject: {best_subject}

Weakest Performing Subject: {weakest_subject}

Students At Risk: {at_risk_students}

Grade Distribution:
{grade_distribution.to_string()}

Subject-wise Average:
{subject_averages.to_string()}
"""

print(report)
```
